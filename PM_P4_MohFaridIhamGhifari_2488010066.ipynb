```python
import pandas as pd
import numpy as np

# Dataset karyawan (sengaja mengandung masalah kualitas)
data = {
    'usia'      : [25, 32, np.nan, 45, 28, 51, 38, np.nan, 29, 41],
    'pendapatan': [4_000_000, 7_500_000, 5_200_000, 12_000_000, 4_800_000,
                   15_000_000, 8_100_000, 6_300_000, np.nan, 9_900_000],
    'pendidikan': ['SMA','S1','SMA','S2','SMA','S2','S1','S1','SMA','S2'],  # ordinal
    'kota'      : ['Bandung','Jakarta','Bandung','Surabaya','Jakarta',
                   'Surabaya','Jakarta','Bandung','Bandung','Jakarta'],    # nominal
    'membeli'   : ['Tidak','Ya','Tidak','Ya','Tidak','Ya','Ya','Tidak','Tidak','Ya']  # target
}
df = pd.DataFrame(data)
```

```python
df['usia'] = df['usia'].fillna(df['usia'].mean())
df['pendapatan'] = df['pendapatan'].fillna(df['pendapatan'].median())
```

```python
df['kelompok_usia'] = pd.cut(
    df['usia'],
    bins=[0, 25, 35, 100],
    labels=['Muda', 'Dewasa', 'Senior']
)
```

```python
X = df.drop(columns=['membeli'])
y = df['membeli']
```

```python
X['pendidikan'] = X['pendidikan'].map({'SMA':0,'S1':1,'S2':2}) # ordinal
X = pd.get_dummies(X, columns=['kota', 'kelompok_usia'], dtype=int) # nominal
y = y.map({'Tidak':0,'Ya':1})
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(
X, y, test_size=0.3, random_state=42)
```

4. Menurut saya, fit_transform digunakan pada data latih karena scaler perlu melihat dan mempelajari nilai dari data latih terlebih dahulu, kemudian melakukan scaling pada data tersebut.
Sedangkan pada data uji cukup menggunakan transform, karena data uji tidak boleh ikut digunakan untuk mempelajari proses scaling. Dengan begitu, data uji tetap menjadi data yang belum digunakan dalam proses pelatihan dan dapat mencegah terjadinya data leakage.
